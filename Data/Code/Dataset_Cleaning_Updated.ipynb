```python
#Import the necessary files
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
#Read the dataset
df= pd.read_csv('ModeChoiceOptima.txt', sep='\t')
```

```python
#To check the format of the dataset
df.head()
```

```python
#Copy the dataframe
df1=df.copy()
```

```python
df1.count()
```

```python
#Keeping all the values across the columns except -2
df1 = df1[(df1 != -2).all(axis=1)]
#Checking the count of entries in the dataset after removing the -2 values
df1.count()
```

```python
#Checking the count of entries in the Choice column with -1 values
df1[df1['Choice'] == -1].count()
```

```python
#Drop the rows with -1 values in the Choice column
df1.drop(df1[df1['Choice'] == -1].index, inplace=True)
#Checking the count of entries after removal of -1 in the Choice column
df1.count()
```

```python
#Converting all the entries with (-1) values to NaN values in the dataset
df1.replace(-1, np.nan, inplace=True)
df1.count()
```

```python
#Filling the NaN values with the mode of the respective columns
df1 = df1.fillna(df1.mode().iloc[0])
```

```python
#Checking the count of entries in the dataset after filling the NaN values with mode of respective columns
df1.count()
```

```python
#Saving the cleaned dataset to a new CSV file
df1.to_csv('Cleaned_Dataset_Updated.csv', index=False)
```

```python
#Summary statistics of the cleaned dataset
df1_stat=df1.describe()
```

```python
#Saving the summary statistics of the cleaned dataset to a new CSV file
df1_stat.to_csv('Summary_Statistics_Cleaned_Dataset_Updated.csv', index=True)
```

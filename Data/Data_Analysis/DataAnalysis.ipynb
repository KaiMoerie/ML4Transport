```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv('ModeChoiceOptima_cleaned.csv')
```

```python
df1=df.copy()
```

```python
df1['age'].value_counts()
```

```python
df1['Gender'].value_counts()
```

```python
df1['Choice'].value_counts()
```

```python
#Choice Labels
choice_labels = {0: 'Public Transport', 1: 'Private (Car)', 2: 'Soft Modes'}
df1['Choice_label'] = df1['Choice'].map(choice_labels)

#Mode Choice Distribution
fig, ax = plt.subplots(figsize=(6,5))
counts = df1['Choice'].value_counts().sort_index()

bars = ax.bar([choice_labels[i] for i in counts.index],
              counts.values,
              color=['#2196F3','#F44336','#4CAF50'],
              edgecolor='white', linewidth=1.5)

for b in bars:
    ax.text(b.get_x()+b.get_width()/2, b.get_height()+10,
            str(int(b.get_height())), ha='center')

ax.set_title('Mode Choice Distribution')
ax.set_ylabel('Count')
ax.spines[['top','right']].set_visible(False)

plt.savefig('plot1_mode_choice.png', dpi=150)
plt.close()
```

```python

fig, ax = plt.subplots(figsize=(6,5))

for mode, color in zip([0,1,2], ['#2196F3','#F44336','#4CAF50']):
    data = df1[df1['Choice']==mode]['age'].dropna()
    ax.hist(data, bins=20, alpha=0.6,
            color=color, label=choice_labels[mode],
            edgecolor='white')

ax.set_title('Age Distribution by Mode')
ax.set_xlabel('Age')
ax.set_ylabel('Count')
ax.legend()
ax.spines[['top','right']].set_visible(False)

plt.savefig('plot2_age.png', dpi=150)
plt.close()
```

```python
#CarAvail Plots
fig, ax = plt.subplots(figsize=(6,5))

caravail_labels = {1: 'Always', 2: 'Sometimes', 3: 'Never'}
cross = df.groupby(['CarAvail','Choice']).size().unstack(fill_value=0)

cross.index = [caravail_labels.get(i, i) for i in cross.index]
cross.columns = [choice_labels[c] for c in cross.columns]

cross_pct = cross.div(cross.sum(axis=1), axis=0) * 100
cross_pct.plot(kind='bar', ax=ax)

ax.set_title('Car Availability vs Mode (%)')
ax.set_ylabel('Share (%)')
ax.spines[['top','right']].set_visible(False)

plt.savefig('plot3_caravail.png', dpi=150)
plt.close()
```

```python
#Envir05 Plots
fig, ax = plt.subplots(figsize=(6,5))

for mode, color in zip([0,1,2], ['#2196F3','#F44336','#4CAF50']):
    sub = df1[(df1['Choice']==mode) & (df1['Envir05'].between(1,5))]['Envir05']
    counts = sub.value_counts().sort_index()
    ax.plot(counts.index, counts.values, marker='o',
            color=color, label=choice_labels[mode])

ax.set_title('Envir05 by Mode')
ax.set_xlabel('Response')
ax.set_ylabel('Count')
ax.legend()
ax.spines[['top','right']].set_visible(False)

plt.savefig('plot5_envir05.png', dpi=150)
plt.close()
```

```python
#Distance Distribution by Mode
fig, ax = plt.subplots(figsize=(6,5))

for mode, color in zip([0,1,2], ['#2196F3','#F44336','#4CAF50']):
    data = df[df['Choice']==mode]['distance_km'].dropna()
    data = data[data < 200]
    ax.hist(data, bins=25, alpha=0.6,
            color=color, label=choice_labels[mode],
            edgecolor='white')

ax.set_title('Trip Distance by Mode')
ax.set_xlabel('Distance (km)')
ax.legend()
ax.spines[['top','right']].set_visible(False)

plt.savefig('plot4_distance.png', dpi=150)
plt.close()
```

```python
#Urban vs Rural Mode Share
fig, ax = plt.subplots(figsize=(6,5))

urb_labels = {1:'Urban', 2:'Rural'}
cross = df.groupby(['UrbRur','Choice']).size().unstack(fill_value=0)

cross.index = [urb_labels.get(i,i) for i in cross.index]
cross.columns = [choice_labels[c] for c in cross.columns]

cross_pct = cross.div(cross.sum(axis=1), axis=0) * 100
cross_pct.plot(kind='bar', ax=ax)

ax.set_title('Urban vs Rural Mode Share')

plt.savefig('plot9_urban.png', dpi=150)
plt.close()
```

```python
#Household Cars vs Mode
fig, ax = plt.subplots(figsize=(6,5))

cross = df[df['NbCar']<=4].groupby(['NbCar','Choice']).size().unstack(fill_value=0)
cross.columns = [choice_labels[c] for c in cross.columns]

cross_pct = cross.div(cross.sum(axis=1), axis=0) * 100
cross_pct.plot(kind='bar', ax=ax)

ax.set_title('Household Cars vs Mode')

plt.savefig('plot10_cars.png', dpi=150)
plt.close()
```

```python
#Trip Purpose vs Mode
fig, ax = plt.subplots(figsize=(6,5))

purpose_labels = {1:'Work', 2:'Work+Leisure', 3:'Leisure'}
cross = df.groupby(['TripPurpose','Choice']).size().unstack(fill_value=0)

cross.index = [purpose_labels.get(i,i) for i in cross.index]
cross.columns = [choice_labels[c] for c in cross.columns]

cross_pct = cross.div(cross.sum(axis=1), axis=0) * 100
cross_pct.plot(kind='bar', ax=ax)

ax.set_title('Trip Purpose vs Mode')

plt.savefig('plot12_purpose.png', dpi=150)
plt.close()
```
